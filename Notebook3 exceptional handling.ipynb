## Exception Handling

### 1. try

**Definition**

The try block is used for exception handling.

It lets you run code that "might cause an error", and then handle that error gracefully instead of crashing the program.

When Python finds an error inside the try block, it stops executing the remaining statements in that block and looks for a suitable except block to handle the error.

Using try does not mean that an error will never happen. It simply gives us a way to handle possible errors without suddenly stopping the whole program.

Example:

Suppose we ask the user to enter two numbers and divide the first number by the second number.

If the user enters 0 as the second number, Python will give a division-by-zero error. We can put the division code inside try so that we can handle this situation.

**Real-Time Example**

Imagine an ATM. When a user enters an invalid amount or performs an operation that cannot be completed, the ATM should show an appropriate message instead of completely crashing. The risky operation can be handled safely using exception handling.

```python
# The code inside the try block is the code that might produce an error.
try:
    number1 = int(input("Enter first number: "))
    number2 = int(input("Enter second number: "))

    result = number1 / number2

    print("Result:", result)           #If an error occurs, Python will look for an except block to handle it.
```

### 2. except

**Definition**

The except block is used to handle an error that occurs inside the try block.

If code inside try block raises an error, Python jumps into the matching except block instead of crashing.

Instead of allowing Python to stop the program and display an error message, we can catch the error and show our own message or perform another action.

We can have different except blocks for different types of errors.

Example:

If we try to divide a number by zero, Python raises a ZeroDivisionError. We can catch that error using except.

**Real-Time Example**

In a login system, a user may enter incorrect information. Instead of allowing the application to crash, the system can catch the problem and display a message such as "Invalid login details."

```python
try:
    number1 = 10   
    number2 = 0     # here num2 is 0 so Python raises a ZeroDivisionError.

    result = number1 / number2

    print(result)

except ZeroDivisionError:       #The except ZeroDivisionError block catches that error and prints a user-friendly message.
    print("Cannot divide by zero.")       #this is our own msg as a reply for the error
```

### 3. else

**Definition**

The else block is used with try and except.

The else block is executed only when the code inside the try block runs successfully without producing an exception.

It is useful when we want to keep the successful part of the operation separate from the error-handling code.

Example:

If two numbers are entered correctly and the division works, we can use else to print the result.

If an error occurs, the else block will not execute.

**Real-Time Example**

When making an online payment, the payment operation can be attempted first. If it is completed successfully, we can show a payment-success message. If something goes wrong, the except block can handle the failure.

```python
try:
    number1 = 20
    number2 = 5      # here error will not raise 

    result = number1 / number2

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:                        #so else block will run 
    print("Division successful.")
    print("Result:", result)
```

### 4. finally

**Definition**

The finally block contains code that should run whether an exception occurs or not.

It is commonly used for cleanup work, such as closing a file, closing a database connection, or releasing a resource.

The important point is that finally normally executes whether the try block succeeds or an exception is handled.

Example:

Suppose we open a file and read its contents. After finishing the operation, we want to make sure the file is closed whether the reading was successful or an error occurred.

**Real-Time Example**

When using a database, a program may open a database connection, perform some operation, and then close the connection. The connection should be closed even if an error happens during the operation.

```python
try:                                  #The division works successfully, so the except block is skipped.
    number = 10 / 2
    print("Result:", number)

except ZeroDivisionError:
    print("Something went wrong.")

finally:               #After that, the finally block runs.
    print("This block always runs.")

#Even if we change the code to cause a ZeroDivisionError, the finally block will still execute.
```

### 5. raise

**Definition**

raise is used to manually trigger an exception in your code.

The raise statement is used when we want to create an exception ourselves.

Normally, Python raises exceptions automatically when something goes wrong. With raise, we can intentionally stop an operation when a condition is not acceptable.

This is useful when we want to apply our own rules to the program.

Example

Suppose a user enters a negative age. Python will not automatically consider this a programming error, but according to our application rules, an age cannot be negative. We can use raise to create an exception ourselves.

**Real-Time Example**

During account registration, if a user enters an age below 0, the application can reject the value and raise an error saying that the age is invalid.

```python
age = -5

if age < 0:   #The program checks whether age is less than 0.
    raise ValueError("Age cannot be negative.")     #here this raise statement manually creates a ValueError. 

print("Age:", age)    #The raise statement manually creates a ValueError.
```

### 6. assert

**Definition**

assert is used to check whether a condition is true.

If the condition is True, the program continues normally.

If the condition is False, Python raises an AssertionError.

assert is like saying “this must be true, or else stop right here.”

It is mainly useful for checking assumptions while developing and testing a program.

Example

Suppose we expect a person's age to be greater than or equal to zero. We can use assert to check that assumption.

**Real-Time Example**

While testing a banking application, we may expect an account balance to never be negative after a particular operation. An assertion can be used during development to check whether this assumption is still true.

```python
x = 10
assert x > 0, "x must be positive"   #here the condition is true so assertionerror will not raise
print("All good!")

y = -5
assert y > 0, "y must be positive"   #here the condition is false so it Raises AssertionError
```

### 7. Custom Exceptions

**Definition**

A custom exception is an exception that we create ourselves for a specific situation in our program.

A custom exception is just a class that inherits from Exception

Python already provides many built-in exceptions such as ValueError, TypeError, and ZeroDivisionError.

But sometimes the built-in exceptions are not descriptive enough for our application. In that case, we can create our own exception class.

Example

Suppose we are creating a bank application and want to show a special error when someone tries to withdraw more money than they have in their account.

We can create our own exception called InsufficientBalanceError.

**Real-Time Example**

In a banking application, if a customer has ₹5,000 but tries to withdraw ₹10,000, the application can raise a custom InsufficientBalanceError instead of showing a general error.

```python
class InsufficientBalanceError(Exception):     # here InsufficientBalanceError is our own exception class.
    pass                                       #It inherits from Python's built-in Exception class.


balance = 5000
withdraw_amount = 10000

try:
    if withdraw_amount > balance:   #here If it is greater, we use raise to create our custom exception.
        raise InsufficientBalanceError("Not enough balance.")

    balance -= withdraw_amount
    print("Withdrawal successful.")

except InsufficientBalanceError as error:  #The except block catches InsufficientBalanceError and displays the message.
    print(error)
```

### 8. ValueError

**Definition**

ValueError is a built‑in exception.

ValueError occurs when we give a function or operation a value that has the correct general type, but the value itself is not acceptable for that operation.

In simple words, Python understands what type of data we gave, but it cannot use that particular value for the operation.

**When It Occurs**

It commonly happens when we try to convert an invalid value into another type.

common cases:

- Converting invalid strings to numbers  
   ValueError: invalid literal for int() with base 10: 'abc'

- Math function with invalid value  
  ValueError: math domain error

- Invalid step in range()  
  ValueError: range() arg 3 must not be zero

- Invalid date in datetime  
  ValueError: month must be in 1..12

- Removing a non‑existent item from a list  
  ValueError: list.remove(x): x not in list


For example, int("hello") causes a ValueError because "hello" is a string, but it cannot be converted into an integer.

Example

If a program asks the user to enter their age and they enter "twenty" instead of 20, converting that input to an integer will cause a ValueError.

**Real-Time Example**

In a registration form, if the user enters "twenty" in an age field where the program expects a number, the program can handle the ValueError and ask the user to enter a valid number.

```python
try:
    age = int("twenty")  #here "twenty" cannot be converted into an integer, Python raises ValueError.
    print("Age:", age)

except ValueError:  #The except ValueError block catches the error and displays a more understandable message.
    print("Please enter a valid number.")
```

```python
# another example with rasie statement
try:
    def set_age(age):
        if age < 0:
            raise ValueError("Age cannot be negative!")  #here with the help of raise we are manually triggering ValueError
        print("Age set to:", age)
    
    set_age(25)   # here this call works fine
    set_age(-5)   # here it Raises ValueError
except ValueError as e:
    print("not valid",e)
```

### 9. TypeError

**Definition**

TypeError is a built‑in exception.

TypeError occurs when we try to perform an operation using an inappropriate data type.

In simple words, the value may be valid, but its data type is not suitable for the operation we are trying to perform.

**When It Occurs**

It commonly happens when we use incompatible data types together.

- String + Integer concatenation  
  TypeError: can only concatenate str (not "int") to str

- Passing wrong type to math function  
  TypeError: must be real number, not str

- Calling a non‑callable object  
  TypeError: 'int' object is not callable

- Multiplying a list by a string  
  TypeError: can't multiply sequence by non-int of type 'str'

- Using len() on an integer  
  TypeError: object of type 'int' has no len()


Example:

If we try to add "10" and 5, Python cannot directly perform the addition because one value is a string and the other is an integer.

**Real-Time Example**

Suppose a billing program receives the product price as text and the quantity as a number. If we try to perform an operation without converting the price to the correct type, a TypeError can occur.

```python
import math
math.sqrt("16")   # expects a number but got a string so it raises an typeError
```

```python
try:
    price = "100"  #  price is a string and quantity is an integer.
    quantity = 2

    total = price + quantity  #Python cannot use + to directly add these two different types.

    print(total)

except TypeError:  #so it rasies a Type Error
    print("Price and quantity must have compatible data types.")
```

### 10. IndexError

**Definition**

IndexError is a built‑in exception.

IndexError occurs when we try to access an index that does not exist in a sequence such as a list, tuple, or string.

Python indexing starts from 0, so we need to make sure the index exists before accessing it.

**When It Occurs**

For example, if a list contains four elements, its valid indexes are 0, 1, 2, and 3.

Trying to access index 4 will cause an IndexError.

like:

- List index out of range  
  IndexError: list index out of range

- Tuple index out of range  
  IndexError: tuple index out of range

- String index out of range  
  IndexError: string index out of range

- Negative index too far  
  IndexError: list index out of range

- Accessing element from an empty list  
  IndexError: list index out of range

Example:

Suppose we have a list containing three fruits. Trying to access the fourth position will cause an error because that position does not exist.

**Real-Time Example**

Imagine a student list containing 30 students. If a program tries to access student number 31 using a list index, it will cause an IndexError because that position does not exist.

```python
items = ["apple", "banana"]
print(items[2])   # here Only 0 and 1 exist so it raises an indexerror
```

```python
try:
    fruits = ["Apple", "Banana", "Mango"]  #here The list contains only three elements so the valid indexes are 0, 1, and 2.

    print(fruits[5])   #Index 5 does not exist, so Python raises IndexError.

except IndexError:
    print("The requested position does not exist.")  #This except block catches the error and prints a meaningful message.
```

### 11. KeyError

**Definition**

KeyError is a built‑in exception.

KeyError occurs when we try to access a key that does not exist in a dictionary.

A dictionary works using key-value pairs, so the key we are trying to access must actually be present.

**When It Occurs**

If a dictionary contains "name" and "age" but we try to access "salary", Python raises a KeyError because "salary" is not available as a key.

- Accessing a missing key in a dictionary
  

- Using .pop() on a non‑existent key


- Accessing a dictionary key directly that doesn’t exist


- Iterating and trying to access a missing key


- Nested dictionary missing a key


Example:

Suppose we have student information containing a name and age. If we try to access the student's email when no email key exists, a KeyError can occur.

**Real-Time Example**

In an employee management system, if the program tries to access an employee's salary using a "salary" key but that information was not stored, a KeyError can occur.

```python
data = {"name": "Hari", "age": 25}
print(data["city"])   # here "city" key doesn’t exist so python raises keyerror
```

```python
try:
    student = {
        "name": "Hari",     #here the dictionary contains only "name" and "age".
        "age": 22
    }

    print(student["salary"])  # so When Python tries to access student["salary"], it raises KeyError

except KeyError:
    print("The requested key is not available.")
```

### 12. AttributeError

**Definition**

AttributeError is a built‑in Python exception.

AttributeError occurs when we try to access an attribute or method that does not exist for an object.

Python understands the object, but the particular property or method we are asking for is not available on that object.

In simple terms: you’re asking an object to use a property or function it doesn’t have.

**When It Occurs**

It can happen when we misspell a method name or try to use a method that belongs to another data type.

- Calling a non‑existent method on a string
  AttributeError: 'str' object has no attribute 'push'

- Accessing a missing attribute in a class object
  AttributeError: 'Person' object has no attribute 'age'

- Using the wrong method for a list
  AttributeError: 'list' object has no attribute 'upper'

- Accessing an attribute before it’s defined
  AttributeError: 'Car' object has no attribute 'speed'

- Misspelling an attribute name
  AttributeError: 'str' object has no attribute 'lowerr'

Example:

A string has an upper() method, but it does not have a method called uppercase().

**Real-Time Example**

Suppose an application has an employee object with a name attribute. If the program tries to access employee.salary when salary was never defined, an AttributeError can occur.

```python
try:
    name = "Hari"

    print(name.uppercase())  #There is no uppercase() method for strings.

except AttributeError:    #Therefore, Python raises AttributeError.
    print("The requested method or attribute does not exist.")
```

```python
#Calling a non‑existent method on a string
text = "hello"
text.push("world")   #here strings don’t have push()
```

```python
#Accessing a missing attribute in a class object
class Person:
    def __init__(self, name):
        self.name = name

p = Person("Hari")
print(p.age)   #here no attribute 'age'
```

```python
#Using the wrong method for a list

nums = [1, 2, 3]
nums.upper()   # lists don’t have upper()
```

```python
#Accessing an attribute before it’s defined

class Car:
    pass

c = Car()
print(c.speed)   # speed not defined yet
```

### 14. FileNotFoundError

**Definition**
FileNotFoundError is an built-in Exception

FileNotFoundError occurs when Python tries to open a file that cannot be found at the specified location.

This usually happens when the filename is wrong, the file has been deleted, or the file is located in a different folder.

**When It Occurs**

If we use open() with a filename that does not exist, Python raises FileNotFoundError.

- Trying to open a file that doesn’t exist
 
- Wrong file name used

- Wrong file path specified

- File deleted before access

- Relative path issue (file not in expected folder)

Example:

Suppose we want to read students.txt, but that file does not exist in our current folder.

**Real-Time Example**

A data analysis program may try to read a CSV file containing customer information. If the file was moved or deleted, the program can handle FileNotFoundError and tell the user that the file could not be found.

```python
file = open("students.txt", "r")   #here the file doest'nt exists.so filenotfounderror is triggered

data = file.read()

file.close()
```

```python
try:
    file = open("students.txt", "r")  #here we r trying to open students.txt file which does'nt exists

    data = file.read()                  #If the file does not exist, open() raises FileNotFoundError.

    file.close()

    print(data)

except FileNotFoundError:  
    print("The file was not found.")
```

Trying to calculate 10 / 0 will cause a ZeroDivisionError.

**Real-Time Example**

```python
x = 10 / 0 
```

```python
x = 7 % 0
```

### 17. ModuleNotFoundError

**Definition**

A subclass of ImportError (introduced in Python 3.6).

Raised "specifically" when the module itself cannot be located.

ModuleNotFoundError occurs when Python cannot find the module that we are trying to import.

It is commonly seen when the module name is incorrect or the required package has not been installed in the current Python environment.

ModuleNotFoundError → Python cannot find the module itself.

ImportError → Python finds the module, but the particular thing we are trying to import is not available.

**When It Occurs**

For example, if we write import mypackage but there is no module or package with that name available, Python raises ModuleNotFoundError.

Example:

We try to import a package that is not installed or does not exist.

**Real-Time Example**

Suppose we create a Python project that needs the pandas library. If pandas is not installed in the project's virtual environment and we try to import it, Python can raise ModuleNotFoundError.

```python
import my_unknown_module
```

```python
try:
    import my_unknown_module #Since that module does not exist in the available Python environment, Python raises ModuleNotFoundError.

except ModuleNotFoundError:
    print("The module was not found. Check the module name or install it.")
```
