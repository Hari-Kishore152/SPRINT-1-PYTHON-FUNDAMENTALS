## Object-Oriented Programming

### 1. Class

**Definition**

In Python, a class is a blueprint for creating objects. 

It defines the attributes (variables) and methods (functions) that the objects created from the class will use.

It defines what kind of data an object can have and what actions it can perform.

A class can contain variables to store data and methods to perform operations on that data.

We create objects from a class.

Example:

Suppose we want to represent students in a Python program. Instead of separately writing the details and functions for every student, we can create a Student class and create different student objects from it.

Real-Time Example:

A college can have a common student structure containing details such as name, age, roll number, and course. Each individual student can then be represented as an object created from that common structure.

```python
class Student:  #Here, Student is the class.
    def __init__(self, name, marks): # here this is called as constructor 
        self.name = name        # attribute
        self.marks = marks      # attribute

    # this is method it defines behavior obj ,what the obj can do or how it should act.
    def display(self):
        print(f"{self.name} scored {self.marks} marks")

# Create objects (instances of Student class)
s1 = Student("Hari", 85)
s2 = Student("Rahul", 90)

# Use methods
s1.display()
s2.display()
```

### 2. Object

**Definition**

An object in Python is an instance of a class.

A class is the blueprint (design).

An object is the actual thing which is created from that blueprint.

Each object has its own attributes (data) and can use the methods (functions) defined in the class.

We can create multiple objects from the same class, and each object can have its own data.

Example:

If Student is a class, then student1 and student2 can be two different objects created from that class.

Both objects follow the same structure, but they can contain different student information.

Real-Time Example:

If Car is a class, then individual cars such as a Honda car or a Toyota car can be considered objects created using that class. They can have different colors, models, and registration numbers.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name      # attribute
        self.marks = marks    # attribute

    def display(self):        # method
        print(f"{self.name} scored {self.marks} marks")


s1 = Student("Hari", 85)  #here s1 is the first obj 
s2 = Student("Rahul", 90) # s2 is the second obj


s1.display()
s2.display()
```

### 3. Constructor (__init__)

**Definition**

A constructor is a special method that is automatically called when we create an new object.

In Python, a constructor is a special method inside a class called __init__().

It automatically runs when you create a new object from the class.

Its job is to initialize the object’s attributes (variables) with values.

The constructor (__init__) takes the values you pass when creating the object.

It assigns those values to the object’s variables (attributes).

So each object gets its own set of variables filled with the values you gave at creation time.


Example:

Suppose we create a Student object. We can use __init__() to automatically assign the student's name and age when creating the object.

Instead of creating the object first and assigning the values separately, we can provide the values while creating the object itself.

Real-Time Example:

When a new employee joins a company, details such as name, employee ID, and department are recorded when the employee profile is created. The __init__() method works in a similar way by setting the initial values of an object.

```python
class Student:
    # Constructor
    def __init__(self, name, marks):
        self.name = name      # attribute
        self.marks = marks    # attribute

    def display(self):        
        print(f"{self.name} scored {self.marks} marks")

# Creating objects
s1 = Student("Hari", 85)   # constructor runs here
s2 = Student("Rahul", 90)  # constructor runs here

s1.display()
s2.display()

# Create object → s1 = Student("Hari", 85)

# Python calls __init__() automatically.

# "Hari" is stored in self.name.

# 85 is stored in self.marks.

# Object ready → Now s1 has its own attributes (name, marks).

# Call method → s1.display() prints the data stored in that object.
```

### 4. Instance Variables

**Definition**

An instance variable in Python is a variable that is defined inside a class but belongs to a specific object (instance).

Each object created from the class gets its own copy of the instance variables, so different objects can hold different values.

They are usually created using self inside the __init__() method.


Example

If we create two students, each student can have a different name and age.

The variable name exists for both objects, but the value stored in it can be different for each object.

Real-Time Example

In a company, every employee has personal information such as name, employee ID, and salary. These values are different for each employee, so they can be stored as instance variables.

```python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary


employee1 = Employee("Hari", 30000)
employee2 = Employee("Rahul", 40000)

print(employee1.name)                     #self.name and self.salary are instance variables.
print(employee1.salary)                   #employee1 and employee2 have their own values.

print(employee2.name)                        
print(employee2.salary)
```

### 5. Class Variables

**Definition**

A class variable is a variable that belongs to the class rather than to a particular object.

Unlike instance variables (which are unique to each object), class variables belong to the class itself, and every object created from that class can access the same value

The same class variable is shared by all objects of that class.

It is useful when we have a value that should be common to all objects.

- Instance variable - belongs to one object, unique per instance.

- Class variable - belongs to the class, shared by all instances.

Example:

Suppose all students belong to the same college. Instead of storing the college name separately for every student, we can create one class variable and share it among all student objects.

Real-Time Example:

If all employees in a company belong to the same company, the company name can be stored as a class variable because it is common to all employees.

```python
class Employee:

    company = "ABC Technologies"   # here company is a class variable because it is defined directly inside the class.

    def __init__(self, name):
        self.name = name


employee1 = Employee("Hari")
employee2 = Employee("Rahul")

print(employee1.name, employee1.company)    #Both employee1 and employee2 can access the same company value.
print(employee2.name, employee2.company)
```

### 6. Instance Methods

**Definition**

An instance method in Python is a function defined inside a class that operates on objects (instances) of that class.

It always takes self as the first parameter, which refers to the specific object calling the method. 

Through self, the method can access and modify the object’s instance variables.

Example

Suppose a student has a method called display_details(). It can use self to access that student's name and age.

The same method can work differently for different student objects because each object has its own data.

Real-Time Example

A bank account can have an instance method such as deposit(). Each bank account object can use the same method, but the money will be added to the particular account that called the method.

```python
class BankAccount:

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):     #here deposit() is an instance method
        self.balance += amount     #It uses self to access the balance of the particular account.
        print("Updated Balance:", self.balance)


account1 = BankAccount("Hari", 10000)

account1.deposit(2000)
```

### 7. Static Methods

**Definition**

A static method in Python is a method inside a class that does not depend on the object (instance) or the class itself.

Unlike instance methods (which use self) and class methods (which use cls), a static method is just a regular function placed inside a class for organization. 

The method behaves like a normal function, but it’s grouped inside the class for better structure.

It does not require self or cls as its first parameter.

We create a static method using the @staticmethod decorator.

- Instance methods → need self, work with object data.

- Class methods → need cls, work with class-level data.

- Static methods → don’t need either, just utility functions inside a class.

  
Example:

Suppose we have a class called Calculator. Addition is a simple operation that does not need any object data, so we can create it as a static method.

Real-Time Example:

A company application may have a utility function to check whether a number is valid. The function does not need employee or company information, so it can be kept as a static method.

```python
class MathUtils:
    # Static method
    @staticmethod
    def add(x, y):
        return x + y

    @staticmethod
    def multiply(x, y):
        return x * y

# Call static methods
print(MathUtils.add(10, 5))       # You can call MathUtils.add(10, 5) directly without creating an object.
print(MathUtils.multiply(4, 3))    
```

### 8. Class Methods

**Definition**

A class method is a method that works with the class itself rather than with a particular object.

It always takes cls as the first parameter, which refers to the class itself. This means class methods can access and modify class variables (shared across all objects), but not instance-specific data unless passed in.

We create a class method using the @classmethod decorator.

The first parameter of a class method is usually cls, which refers to the class.

in short:
- Instance methods = behavior of individual objects.

- Class methods = behavior of the class as a whole.

- Static methods = general utility functions grouped inside the class.

Example

Suppose we want to keep track of the total number of students created from a class. A class method can access and modify the class-level value.

Real-Time Example

A company may want to keep track of the total number of employees registered in its system. Since the total belongs to the company rather than one employee, a class method can be used to work with that information.

```python
class Student:
   
    school_name = "ABC School"  # here this is Class variable

    def __init__(self, name, marks):
        self.name = name        # instance variable
        self.marks = marks      # instance variable

                                       
    @classmethod
    def change_school(cls, new_name):    # this is Class method
        cls.school_name = new_name

# Create objects
s1 = Student("Hari", 85)
s2 = Student("Rahul", 90)

print(s1.school_name)   
print(s2.school_name)   

# Change class variable using class method
Student.change_school("XYZ School")

print(s1.school_name)   
print(s2.school_name)   
```

### 9. Properties (@property)

**Definition**

A property allows us to access a method like a normal variable.

A property in Python is a special way to control access to instance variables.

It lets you get, set, or delete an attribute using methods, but still access it like a normal variable.

It is created using the @property decorator.

Properties are useful when we want to control how a value is accessed or calculated without changing the way we use it.

Example:

Suppose we have a person's first name and last name. Instead of storing the full name separately, we can create a property that combines both values and gives us the full name whenever we access it.

Real-Time Example:

In an employee system, we may store first name and last name separately but want to display the employee's full name whenever needed. A property can calculate and return the full name while allowing us to access it like a normal variable.

```python
class Employee:

    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    @property
    def full_name(self):     #full_name() is actually a method, but because of @property, we can access it like a normal variable.
        return self.first_name + " " + self.last_name


employee = Employee("Hari", "Kishore")

print(employee.full_name)     #We use employee.full_name instead of employee.full_name().
```

```python
# another example with detailed explanation
class Student:
    def __init__(self, name, marks):
        self._name = name          # instance variable
        self._marks = marks        # instance variable (protected)

    # Getter method
    @property
    def marks(self):            ##@property makes marks() act like a variable (s1.marks) instead of a method call.
        return self._marks

    # Setter method
    @marks.setter
    def marks(self, value):                   
        if value < 0 or value > 100:
            print("Invalid marks! Must be between 0 and 100.")
        else:
            self._marks = value

    # Deleter method
    @marks.deleter
    def marks(self):
        print("Deleting marks...")
        del self._marks

# Create object
s1 = Student("Hari", 85)

# Access property like a variable
print(s1.marks)     # 85  #The setter runs when you assign s1.marks = value.

# Update property with validation
s1.marks = 95       # valid          #The getter runs when you access s1.marks.
print(s1.marks)     # 95             #The deleter runs when you delete del s1.marks.


s1.marks = 120      # invalid → "Invalid marks!"
```

### 10.Encapsulation

**Definition**

Encapsulation means keeping the data(variables) and the methods that work on that data together inside a class.

It also helps us control how the data can be accessed or changed from outside the class.

In Python, we commonly use naming conventions like _variable and __variable to indicate protected and private data.

Example

Suppose we have a bank account. The balance should not be changed directly from outside the class. Instead, we can provide methods such as deposit() and withdraw() to control how the balance is changed.

Real-Time Example

In a banking application, the account balance is sensitive information. Users should not directly change the balance. The application provides specific operations like deposit and withdrawal to control the balance.

```python
class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance


account = BankAccount(10000)

account.deposit(2000)

print(account.get_balance())
```

### 11.Abstraction

**Definition**

Abstraction means hiding unnecessary implementation details and showing only the important part to the user.

The user only needs to know what an operation does, not all the internal steps used to perform it.

In Python, abstraction can be implemented using abstract classes and abstract methods.

simply abtsraction allows us to focus on what to do rather than how to do.

Example

When we use a function such as print() or len(), we don't need to know all the internal code that Python uses to perform the operation.

We simply use the function and get the required result.

Real-Time Example

When we use an ATM, we select options such as checking balance or withdrawing money. We don't need to know how the bank's internal systems process the transaction. The complicated implementation is hidden from us.

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car engine started!")

class Bike(Vehicle):
    def start(self):
        print("Bike engine started!")

# Using abstraction
v1 = Car()
v2 = Bike()

v1.start()
v2.start()   

#Vehicle is an abstract class — it defines the idea of a vehicle but doesn’t implement details.

#start() is an abstract method — subclasses must implement it.

#Car and Bike provide their own implementations.

#The user only calls start() without worrying about how each vehicle starts internally.
```

### 12. Inheritance

**Definition**

Inheritance allows one class to use the properties and methods of another class.

The existing class is called the parent or base class, and the new class is called the child or derived class.

It helps us reuse existing code instead of writing the same code again.

Example:

Suppose we have a Vehicle class with a method called start(). A Car class can inherit from Vehicle and use the same start() method.

Real-Time Example:

A car and a bike are both vehicles. They can share common features such as starting and stopping. Instead of defining these common features again, they can be inherited from a common Vehicle class.

```python
class Vehicle:

    def start(self):
        print("Vehicle started")


class Car(Vehicle):

    def drive(self):
        print("Car is moving")


car = Car()

car.start()
car.drive()

#Vehicle is the parent class.

# Car is the child class and inherits from Vehicle.

# Because of inheritance, the Car object can use the start() method from the Vehicle class.

# It can also have its own method, such as drive().
```

### 13. Multiple Inheritance

**Definition**

Multiple inheritance means that one child class inherits from more than one parent class.

This allows the child class to use features from multiple parent classes.

Example

Suppose one class provides printing functionality and another class provides scanning functionality. A PrinterScanner class can inherit from both classes.

Real-Time Example

A multifunction printer can perform more than one task, such as printing and scanning. In Python, a class can inherit features from multiple classes in a similar way.

```python
class Printer:

    def print_document(self):
        print("Printing document")

                                             ## Printer and Scanner are two parent classes.
class Scanner:

    def scan_document(self):
        print("Scanning document")


class PrinterScanner(Printer, Scanner):    ## PrinterScanner inherits from both classes.
    pass


device = PrinterScanner()

device.print_document()
device.scan_document()   # Therefore, its object can use methods from both Printer and Scanner.


```

### 14. Multilevel Inheritance

**Definition**

Multilevel inheritance happens when a class inherits from another child class, creating a chain of inheritance.

In simple words, we have a parent class, a child class, and then another class that inherits from the child class.

Example:

Suppose Animal is a parent class, Dog inherits from Animal, and Puppy inherits from Dog.

The Puppy can access methods from both Dog and Animal.

Real-Time Example:

A company may have different levels of employees. A general employee can have common features, a manager can inherit those features, and a senior manager can inherit features from the manager.

```python
class Animal: #parent calss 

    def eat(self):
        print("Animal eats")


class Dog(Animal):     #here this is already an child class

    def bark(self):
        print("Dog barks")


class Puppy(Dog):       #this class is inherinting from the child class

    def play(self):
        print("Puppy plays")


puppy = Puppy()    #now this puppy class can access all the methonds from both classes

puppy.eat()
puppy.bark()
puppy.play()
```

### 15. Hierarchical Inheritance

**Definition**

Hierarchical inheritance happens when multiple child classes inherit from the same parent class.

The parent class contains common functionality, while each child class can have its own additional functionality.

Example:

Suppose Vehicle is a parent class. Both Car and Bike can inherit from Vehicle.

Real-Time Example:

A company may have different types of employees such as developers and testers. Both can share common employee information while also having their own specific responsibilities.

```python
class Vehicle:                       #here vehicle is the common parent class.

    def start(self):
        print("Vehicle started")


class Car(Vehicle):                       

    def drive(self):
        print("Car is driving")


class Bike(Vehicle):                  #Both Car and Bike inherit from Vehicle.

    def ride(self):
        print("Bike is riding")


car = Car()              #Therefore, both objects can use the start() method while also having their own methods.
bike = Bike()

car.start()
car.drive()

bike.start()
bike.ride()
```

### 16. Polymorphism

**Definition**

Polymorphism means "many forms."

In Python, the same method or operation can behave differently depending on the object that is using it.

This allows us to use a common interface for different types of objects.

simply it allows methods with same name to behave differently based on the object context.

Example

Suppose both a Dog and a Cat have a method called sound(). The method name is the same, but each class can give a different result.

Real-Time Example

Different animals make different sounds. We can use the same action, such as sound(), but a dog can bark while a cat can meow.

```python
class Dog:

    def sound(self):
        print("Dog barks")


class Cat:

    def sound(self):
        print("Cat meows")


animals = [Dog(), Cat()]

for animal in animals:
    animal.sound()

# Both classes have a method with the same name, sound().

# When the loop calls animal.sound(), Python uses the version belonging to the current object.

# So the Dog object barks and the Cat object meows.
```

### 17 Method Overriding

**Definition**

Method overriding happens when a child class provides its own version of a method that already exists in the parent class.

The child class changes the behavior of that inherited method.

- Same method name in parent and child.

- Child’s version replaces (overrides) the parent’s version when called on the child object.

- Enables polymorphism (different behavior depending on the object).

Example:

Suppose the parent class has a sound() method. The child class can define its own sound() method with different behavior.

Real-Time Example:

A general vehicle may have a start() operation, but different vehicles can start in different ways. A child class can override the parent's method to provide its own behavior.

```python
class Vehicle:

    def start(self):            # here vehicle class has a start() method
        print("Vehicle is starting")


class Car(Vehicle):

    def start(self):       #here the Car class inherits from Vehicle but creates its own start() method.
        print("Car starts with a key")


vehicle = Vehicle()
car = Car()

vehicle.start()
car.start()        #here when car.start() is called, the child class version is executed instead of the parent class version.




```

### 18. Method Overloading (Python Approach)

**Definition**

Method overloading means having the same method name with different numbers or types of parameters.

Python does not support traditional method overloading like some other programming languages like java ,c++.

Instead, we can achieve similar behavior using default arguments, *args, or other techniques.

Example

Suppose we want an add() function that can add two numbers or three numbers. In Python, we can use *args so the same function can accept different numbers of arguments.

```python
#examplpe with default arguments
class MathOps:
    def add(self, a=0, b=0, c=0):
        return a + b + c

m = MathOps()
print(m.add(5, 10))       # Here, one method add() handles different numbers of arguments.
print(m.add(5, 10, 20))   
print(m.add(5))           
```

```python
# example with variablel lenght argumrnts
class Calculator:

    def add(self, *numbers):    #here *numbers allows the add() method to accept different numbers of arguments.

        total = 0

        for number in numbers:
            total += number

        return total


calculator = Calculator()

print(calculator.add(10, 20))
print(calculator.add(10, 20, 30))   #So the same add() method can handle two numbers, three numbers, four numbers, or more.
print(calculator.add(10, 20, 30, 40))
```

### 19. Composition

**Definition**

Composition is a relationship where one class contains an object of another class and strongly depends on it.

or

Composition in Python (and OOP) is the principle of building complex classes by combining simpler ones, rather than inheriting from them.

In simple words, one object is made up of another object. If the main object is removed, the contained object is usually not useful separately in that relationship.

- Inheritance says: “A Dog is an Animal.”

- Composition says: “A Car has an Engine.”

It’s about using one class inside another instead of extending it.

Example:

Suppose we have a Car class and an Engine class. A car needs an engine to work, so the Car class can create and contain an Engine object.

In composition, the child object cannot exist without the parent (Car has an Engine — if the Car is destroyed, the Engine goes too).

Real-Time Example:

A car contains an engine. The engine is an important part of the car, and the car is designed around having an engine. This can be represented using composition.

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()   # The Car class creates an Engine object inside its __init__() method.

    def drive(self):
        self.engine.start()      # here Car doesn’t inherit from Engine. Instead, it contains an Engine object and uses it. 
        print("Car is driving")

c = Car()
c.drive()


#Engine = component (it just provides functionality).

# Car = controller (it manages the Engine and decides when to start it).

# In short: Car controls the Engine, while Engine only provides the service.
```

### 20. Aggregation

**Definition**

Aggregation is a relationship where one class uses an object of another class, but both objects can exist independently.

In simple words, one object contains or uses another object, but it does not completely control its lifetime.

- The child object can exist independently of the parent.

- The parent just uses the child object, it doesn’t own or control its lifecycle.

Example:

Suppose we have a Department class and an Employee class. A department can have employees, but an employee can still exist even if that department is removed.

Real-Time Example:

A college has many teachers. If a particular department is closed, the teachers can still exist and can be moved to another department. This is an example of aggregation.

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self, engine):
        self.engine = engine   # Car *has* an Engine (passed in)

    def start_car(self):
        self.engine.start()
        print("Car started")

# Engine created independently
engine1 = Engine()     #Engine exists independently (engine1).

# Car aggregates the engine
car = Car(engine1)     #Car just uses it by holding a reference.
car.start_car()

#If you delete Car, the Engine object still exists.
```

### 21. Association

Association means that two classes are connected because their objects interact with each other.

Neither object necessarily owns the other. They simply work together to perform some task.

in short:

- Association: A general relationship where two independent objects interact with each other.
- Aggregation: A “has-a” relationship where the contained object can exist independently of the container.
- Composition: A strong “has-a” relationship where the contained object cannot meaningfully exist without the container.

Example:

A Teacher can teach a Student. The teacher and student are two separate objects, but they interact with each other.

Real-Time Example:

A doctor and a patient are separate people. They don't belong to each other, but they interact when the doctor provides treatment to the patient.

```python
class Student:                                 #Student and Teacher are separate classes.


    def __init__(self, name):
        self.name = name


class Teacher:

    def teach(self, student):           #The Teacher method receives a Student object and interacts with it.
        print("Teacher is teaching", student.name)


student = Student("Hari")
teacher = Teacher()

teacher.teach(student)     #The teacher does not create or own the student.



```

### 21. Magic (Dunder) Methods

**Definition**

Magic methods, also called dunder methods, are special methods in Python whose names start and end with double underscores.

They allow us to define how objects behave when Python performs certain operations on them.

They let you customize how your objects behave with built‑in Python operations.

- When you write print(obj), Python internally calls obj.__str__().

- When you write len(obj), Python internally calls obj.__len__().

- When you do obj1 + obj2, Python internally calls obj1.__add__(obj2).

So even though you don’t see the method being called, Python is triggering it automatically.

Some common examples are __init__(), __str__(), __len__(), and __repr__().

Example:

When we use +, print(), or len() with an object, Python can internally call special methods to decide how that object should behave.

Real-Time Example:

A shopping cart can be treated like a normal object, but we may want len(cart) to return the number of products in it. We can use a dunder method such as __len__() to define this behavior.

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):
        return f"Book: {self.title}"

    def __len__(self):
        return self.pages

    def __add__(self, other):
        return self.pages + other.pages

b1 = Book("Python Basics", 200)
b2 = Book("Advanced Python", 300)

print(b1)             # print(b1) works because of __str__. (calls __str__)
print(len(b1))        # len(b1) works because of __len__. (calls __len__)
print(b1 + b2)       # b1 + b2 works because of __add__. (calls __add__)


#Here, the magic methods make Book objects behave like built‑in types:
```

### 22. super()

**Definition**

super() is used to access methods or the constructor of a parent class from a child class.

or

super() is used inside a child class to call methods from its parent class.

It’s especially useful when you override a method **but still want to run the parent’s version.**


Example:

Suppose a parent class has a constructor that sets a name. A child class can use super() to call the parent's constructor and then add its own information.

Real-Time Example:

A general employee class may store common information such as name. A developer class can reuse that information and additionally store programming skills. super() helps the child class use the parent's setup.

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def speak(self):
        super().speak()   # here it is using calling the  parent method
        print("Dog barks")   #so now we can use both parent method along with the overrided method

d = Dog()            #here super().speak() lets us reuse the parent’s method before adding the child’s behavior.
d.speak()
```

```python
# second example
class Employee:

    def __init__(self, name):
        self.name = name


class Developer(Employee):

    def __init__(self, name, skill):
        super().__init__(name)
        self.skill = skill


developer = Developer("Hari", "Python")

print(developer.name)
print(developer.skill)
```

### 23. __str__()

**Definition**

__str__() is a special method used to define a readable string representation of an object.

It’s automatically called when you use print(obj) or str(obj).


Example:

Instead of printing something like an object address, we can use __str__() to display useful information about the object.

**Real-Time Example:**

In a student management system, when we print a student object, we may want to see the student's name and course instead of Python's default object information.

```python
class Student:

    def __init__(self, name, course):
        self.name = name
        self.course = course

    def __str__(self):     #__str__() returns the text that should be displayed when the object is printed.
        return f"{self.name} - {self.course}"


student = Student("Hari", "B.Tech")

print(student)
```

### 24. __repr__()

**Definition**

__repr__() is a special method used to provide a more detailed representation of an object.

It is mainly useful for debugging and development because it should give enough information to understand what the object contains.

It’s meant for developers/debugging, not end‑users.

Called when you use repr(obj) or just type the object name in the Python shell.



Example:

For a student object, we can make __repr__() display the class name along with the student's details.

**Real-Time Example:**

When a developer is checking objects during debugging, a detailed representation can help them quickly understand the data stored inside each object.

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"


student = Student("Hari", 22)

print(repr(student))
#output:
# Student(name='Hari', age=22) This looks like code that could be used to recreate the object — that’s the idea of __repr__().

#The result gives a clear representation containing the class name and the object's data.
```

### 25. __len__()

**Definition**

__len__() is a special method that allows us to define what should be returned when we use the len() function on our object.

It should return the number of items or the size represented by the object.

Example:

If we create a ShoppingCart class, we can use __len__() to return the number of products currently present in the cart.

**Real-Time Example**

An online shopping cart can contain different products. When we check the cart, we may want to know how many products are currently added. len(cart) can provide that count.

```python
class ShoppingCart:

    def __init__(self):
        self.items = []

    def add_item(self, item):      #add_item() adds new products to the list.
        self.items.append(item)

    def __len__(self):            #__len__() returns the number of items in the list.
        return len(self.items)


cart = ShoppingCart()

cart.add_item("Laptop")
cart.add_item("Mouse")
cart.add_item("Keyboard")

print(len(cart))
```

### 26. Dataclasses (Bonus)

**Definition**

A dataclass is a convenient way to create classes that are mainly used for storing data.

Python's dataclasses module automatically provides common methods such as __init__() and a useful representation of the object.

Dataclasses are perfect when you need simple classes to store data without writing repetitive code

In short: Dataclasses = less boilerplate, more readable, auto‑generated methods for data‑holding classes.

It helps us avoid writing repetitive code when creating simple data-based classes.

Example:

Suppose we want to store a student's name, age, and course. With a dataclass, we don't need to manually write the constructor.

**Real-Time Example**

In a student management system, we may need to store many student details such as name, age, roll number, and course. A dataclass makes it easier to create objects for storing this information.

```python
from dataclasses import dataclass


@dataclass   #@dataclass tells Python to treat the class as a data-focused class.
class Student:    #here we don't need to manually write the constructor
    name: str     #Python automatically creates the __init__() method for us.
    age: int
    course: str


student = Student("Hari", 22, "B.Tech")

print(student)          #It also provides a useful string representation of the object.
```

### 27. Singleton Pattern (Bonus)

**Definition**

The Singleton Pattern is a design pattern used when we want a class to have only one object throughout the program.

Instead of creating a new object every time, the same object is reused.

Example:

Suppose we have a class for application settings. We may want the entire application to use one common settings object instead of creating multiple copies.

**Real-Time Example**

A logging system in an application can use a single logger object so that different parts of the application write their messages through the same logging system.

```python
class Singleton:

    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)

        return cls._instance


object1 = Singleton()
object2 = Singleton()

print(object1 is object2)
```
