### 2. Tables

**Definition**

A collection of related data entries stored in rows (records) and columns (fields).

A table is a structured way of storing data inside a database.

A table consists of rows and columns.

each column represents a specific attribute or field (e.g., name, salary).

Each row represents a single record (e.g., one employee).

**Real-Time Example**

In an e-commerce application, a products table can store product ID, product name, price, category, and stock quantity.



```python
#here we r creating a table named students.
CREATE TABLE students (
    student_id INT,
    name VARCHAR(50),
    age INT,
    course VARCHAR(50)
);
```

### 3. Primary Key

**Definition**

A primary key is a special column (or set of columns) in a database table that uniquely identifies each row in that table. 

It ensures that no two rows can have the same value for the primary key, and it prevents duplicate or null entries.

or

A primary key is a column, or combination of columns, that uniquely identifies each row in a table.

A primary key must contain unique values and cannot contain NULL.

**Real-Time Example**

A bank assigns a unique account number or customer ID to identify customers. Similarly, a database can use a primary key to uniquely identify each record.

```python
CREATE TABLE students (
    student_id INT PRIMARY KEY,   #now here student_id is the primary key column
    name VARCHAR(50),
    age INT,
    course VARCHAR(50)
);
```

### 4. Foreign Key

**Definition**

A foreign key is a column (or set of columns) in one table that refers to the primary key in another table. 

It creates a link between the two tables, enforcing relationships and maintaining data integrity.

A foreign key is a column that is used to connect one table with another table.

A foreign key usually refers to the primary key of another table.

The primary key identifies a record in its own table, while the foreign key creates a connection to a record in another table.


**Real-Time Example**

In an online shopping application, an order table can contain a customer_id that refers to the customer who placed that order.

```python
CREATE TABLE students (
    student_id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(50)
);
```

```python
CREATE TABLE enrollments (
    enrollment_id INT PRIMARY KEY AUTO_INCREMENT,
    student_id INT,
    course_name VARCHAR(50),

    FOREIGN KEY (student_id)
        REFERENCES students(student_id)
);
```

### 5. Relationships

**Definition**

A relationship describes how records in one table are connected to records in another table.

In a relational database, tables are usually connected using primary keys and foreign keys.

There are three common types of relationships:

- One-to-One :  An employee may have one company ID card, and that ID card belongs to one employee.
- One-to-Many :  One customer can have many orders in an online shopping application, but each order belongs to one customer.
- Many-to-Many : One student can enroll in many courses.One course can have many students.



```python
#One-to-Many Relationship
CREATE TABLE customers (
    customer_id INT PRIMARY KEY,
    name VARCHAR(50)
);
CREATE TABLE orders (
    order_id INT PRIMARY KEY,
    customer_id INT,
    product VARCHAR(100),

    FOREIGN KEY (customer_id)
        REFERENCES customers(customer_id)
);  

#customer_id in the orders table connects the orders to the customer
```

### 6.Constraints

**Definition**

Constraints are rules that we apply to columns in a table to control what kind of data can be stored.

They prevent invalid or inconsistent data from being entered.

They help maintain accuracy, consistency, and reliability of data in a database.

For example, if a column is defined as NOT NULL, the database will not allow an empty NULL value in that column.

Common SQL constraints are:

- PRIMARY KEY → Ensures each row in a table is uniquely identifiable.

- FOREIGN KEY → Links one table to another by referencing a primary key.

- NOT NULL → Prevents a column from having empty (NULL) values.

- UNIQUE → Guarantees all values in a column are distinct.

- DEFAULT → Assigns a preset value if none is provided during insertion.

- CHECK → Enforces a condition that values in a column must satisfy.

**Real-Time Example**

Think of constraints as rules in a form. For example, a registration form may require a name, prevent duplicate email addresses, and allow age only within a valid range. Database constraints enforce similar rules automatically.

### 3.ORDER BY

**Definition**

The ORDER BY clause in SQL is used to sort the result set of a query in either ascending (ASC) or descending (DESC) order.

ORDER BY is used to sort the result of a query.

Default order is ascending (ASC) unless explicitly set to descending (DESC).

We can sort data:

ASC → Ascending order
DESC → Descending order

### 4.GROUP BY

**Definition**

The GROUP BY clause in SQL is used to arrange rows that have the same values in specified columns into groups, often so you can apply aggregate functions (like SUM, COUNT, AVG) to each group.

GROUP BY is used to group rows that have the same value in one or more columns.

It is commonly used with aggregate functions such as:

COUNT()
SUM()
AVG()
MIN()
MAX()

Example:

Suppose we want to know how many employees are in each department.

SELECT department, COUNT(*)

FROM employees

**Real-Time Example**

### 5.HAVING

**Definition**
The HAVING clause in SQL is used to filter groups of rows created by the GROUP BY clause based on a conditioin.

HAVING is used to filter groups after GROUP BY has been applied.

The easiest way to remember the difference is:

WHERE  is used to  filter rows
HAVING is used to filters groups

example:

Suppose we want only departments having more than one employee.

SELECT department, COUNT(*)

FROM employees

GROUP BY department

HAVING COUNT(*) > 1;

## Joins

### 1. INNER JOIN

**Definition**

INNER JOIN returns only the rows where a matching value exists in both tables.

In other words:

Give me the records that have a match in Table A AND Table B.

If a record doesn't have a matching value in the other table, it is not included in the result.

**Real-Time Example**

If an application wants to display only employees who are assigned to a valid department, an INNER JOIN can be used.

### 2. LEFT JOIN

**Definition**

LEFT JOIN returns all rows from the left table, along with matching rows from the right table.

If a row from the left table does not have a match in the right table, SQL still returns that row, but the columns from the right table will contain NULL.

**The important point is:**

LEFT JOIN keeps everything from the left table.

example:

### 3. RIGHT JOIN

**Definition**

RIGHT JOIN is the opposite of LEFT JOIN.

It returns all rows from the right table, along with matching rows from the left table.

If a row in the right table has no matching row in the left table, the columns from the left table contain NULL.

**The important point is:**

RIGHT JOIN keeps everything from the right table.

**Real-Time Example**

A company may want to see every department, including departments that currently have no employees. A RIGHT JOIN can return all departments and show NULL for departments without employees.

### 4. FULL JOIN

**Definition**

FULL JOIN returns all rows from both tables.

It includes:

- Rows that match in both tables.
- Rows that exist only in the left table.
- Rows that exist only in the right table.

Where there is no matching record, SQL places NULL for the missing side.

**The important point is:**

FULL JOIN keeps everything from both tables.

**Real-Time Example**

Suppose we want to compare two datasets and identify both matching and unmatched records from both sides. A FULL JOIN can be useful for this type of comparison.

**MySQL does not directly support FULL OUTER JOIN as a normal JOIN keyword so we use union in B/W left join and the right join**

SELECT employees.name,
departments.department_name
FROM employees
LEFT JOIN departments
ON employees.department_id = departments.department_id

UNION

SELECT employees.name,
departments.department_name
FROM employees
RIGHT JOIN departments
ON employees.department_id = departments.department_id;

### 5. CROSS JOIN

**Definition**

A CROSS JOIN in SQL is used to combine every row from one table with every row from another table, producing the Cartesian product of the two tables.
It does not require a matching condition such as:

**It does not require a matching condition such as:**

ON table1.id = table2.id

- If Table A has 3 rows and Table B has 4 rows:

   3 × 4 = 12 combinations

So the result will contain 12 rows.

**Real-Time Example**

An online clothing store can generate all possible combinations of available colors and sizes before creating product variants.

```python
SELECT colors.color,
       sizes.size
FROM colors
CROSS JOIN sizes;
```

### 2. CASE

**Definition**

CASE is used to apply conditional logic inside an SQL query.

It works somewhat like if, elif, and else in Python.

We can check a condition and return different values depending on whether the condition is true or false.

CASE → Evaluates conditions and returns a value when the first condition is met.

If no condition matches, an ELSE value can be returned.

**Real-Time Example**

A company can use CASE to classify employees based on salary, customers based on spending, products based on price, or students based on marks.

### 3. Subqueries

**Definition**

A subquery is a SQL query written inside another SQL query.

The inner query is called the subquery, and the outer query uses the result produced by that subquery.

A subquery is useful when the result of one query is needed as an input for another query.

simply Subquery is A query inside parentheses that provides data to the main query.

Can be used in SELECT, FROM, WHERE, or HAVING clauses.

**Real-Time Example**

A company can use a subquery to find employees earning more than the average salary, products costing more than the average product price, or students scoring above the average marks.

### 4. Common Table Expressions (CTEs)

**Definition**

A Common Table Expression (CTE) is a temporary named result set that we create using the WITH keyword and then use inside a SQL query.

A CTE allows us to break a complicated SQL query into smaller and more understandable steps.

Instead of putting everything into one large query, we can first create a temporary result and then query that result.

Acts like a temporary table or view, but is not stored permanently.

The basic structure is:

WITH temporary_name AS (
    SELECT ...
)
SELECT ...
FROM temporary_name;

A CTE exists only for the duration of the query. It does not permanently create a table in the database.

**Real-Time Example**

In data analysis, a CTE can be used to first calculate monthly sales, then calculate the average sales, and finally identify months where sales were above average.

### 5. Views

**Definition**

A View is a virtual table created from a SQL query.

A view does not normally store a separate copy of the query result like a regular table. Instead, it stores the SQL query definition, and when we use the view, the database uses that definition to retrieve the data.

A view is useful when we have a query that we need to use repeatedly.

Instead of writing the same complicated query every time, we can create a view once and then query the view like a table.

Table → Stores data

View  → Stores a query definition


**Real-Time Example**

A company may have a large employee table containing sensitive and unnecessary columns. A view can be created that shows only employee name, department, and job title for a reporting team.

### 6. Indexes

**Definition**

An index is a database structure used to make searching and retrieving data faster.

It is built from one or more columns and stored in a B-tree (balanced tree) structure.

Without an index, the database may need to examine many rows to find the records matching a condition.

An index creates an additional structure that helps the database locate rows more efficiently.

The database engine uses this structure to quickly locate rows that match a query condition.

However, indexes are not free. They require additional storage, and operations such as INSERT, UPDATE, and DELETE can require extra work because the index may also need to be updated.

**Real-Time Example**

Imagine a library with thousands of books. If the books are not organized properly, finding one particular book can take a long time. An index is similar to a catalog that helps locate information more efficiently.

Important Point

An index can improve read/search performance, but creating indexes on every column is not a good idea.

### 7. Window Functions

**Definition**

A window function performs a calculation across a group of related rows without combining those rows into a single row.

This is the important difference between a window function and GROUP BY.

With GROUP BY, multiple rows are usually combined into one result row per group.

With a window function, the original rows remain visible, while the calculated value is added as an additional column.

#### ROW_NUMBER()

**Definition**

ROW_NUMBER() assigns a unique sequential number to every row within a window.

The numbering starts from 1.

If we use:

**PARTITION BY department**

the numbering starts again from 1 for each department.

#### RANK()

**Definition**

RANK() assigns a rank to rows based on the specified ordering.

If two or more rows have the same value, they receive the same rank.

When there is a tie, the next rank is skipped.

For example:

Salary
70000 → Rank 1
70000 → Rank 1
60000 → Rank 3
50000 → Rank 4

Notice that there is no rank 2.

#### DENSE_RANK()

**Definition**

DENSE_RANK() is similar to RANK(), but when two or more rows have the same value, the next rank is not skipped.

For example:

Salary
70000 → Rank 1
70000 → Rank 1
60000 → Rank 2
50000 → Rank 3

#### LAG()

**Definition**

LAG() is a window function used to access the value from a previous row without using a self-join.

It is useful when we want to compare the current row with an earlier row.

For example, suppose we have monthly sales:

```python
SELECT
    month,
    sales,
    LAG(sales) OVER (
        ORDER BY month
    ) AS previous_month_sales
FROM monthly_sales;
```

#### LEAD()

**Definition**

LEAD() is the opposite of LAG().

It is used to access the value from a following row without using a self-join.
