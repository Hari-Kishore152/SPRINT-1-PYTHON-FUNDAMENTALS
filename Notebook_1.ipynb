### if / elif / else

**Definition**

`if`, `elif`, and `else` are conditional statements in Python.
they are used when we want the program to make a decision based on a condition.

- "if" checks the 1st condition.
- "elif" checks another condition if the previous condition is false.
- "else" runs when none of the above conditions are true.

In simple words, they help the program decide what to do depending on the situation.

EXAMPLE:

If we have a person's age:

- If the age is 18 or above then the person is eligible to vote.
- Otherwise the person is not eligible.

Now with the help of conditional statements the program checks the condition and chooses the appropriate result.

**Real-World Example**

In a login system:

- If the username and password are correct then allow the user to login.
- If the password is wrong then show "Incorrect password".
- Otherwise ask the user to enter valid details.

This is how `if`, `elif`, and `else` are used to make decisions in real applications.

```python
#now here we are checking whether the user is eligile to vote 
age = 20

if age >= 18:
    print("Eligible to vote")
elif age > 0:
    print("Not eligible to vote")
else:
    print("Invalid age")
```

###  for Loop

 **Definition**

A `for` loop is used when we want to repeat a block of code for each item in a sequence or for a specific number of times.

Instead of writing the same code again and again, we can use a "for" loop to repeat it automatically.

Mainly we use "for" loop when we know how many times we should repeat something or iterate through the items in a collection.


#### Simple Example

Suppose we have a list of fruits and we want to print the three fruit names:

Apple, Banana, Mango
Instead of writing three separate print statements, we can use a "for" loop to go through each fruit and print it.

#### Real-World Example:

### while Loop
**Definition**

A "while" loop is used when we want to repeat a block of code as long as a particular condition remains true.

We will use "for" loop when we know how many times to repeat and we us "while" loop when we dont know how many times we should loop over or repeat something

The condition is checked before every iteration.

simply the loop keeps running while the given condition is true and stops when the condition becomes false.

### Example

Suppose we want to print numbers from 1 to 5.

We can start with `1` and keep increasing the number by `1` while the number is less than or equal to `5`.

### Real World Example

Consider a login system that allows a user to enter a password while the password is incorrect.

The program can continue asking for the password while the entered password is wrong.

Once the correct password is entered, the condition becomes false and the loop stops.

```python
crct_pin = "1234"
pin = input("Enter your PIN: ")
# While loop keeps running until the PIN is correct
while pin != crct_pin:
    print("Incorrect PIN. Try again.")
    pin = input("Enter your PIN: ")

print("Access granted")
```

### break

**Definition**
The "break" statement is a  loop control statement.

The "break" statement is used to immediately stop a loop when certain condition is met.

When Python reaches "break", it comes out of the loop even if the loop condition is still true. 


**Real World Example**
In an search system, if we are looking for a particular product.

Once the product is found, the program should stop searching instead of checking every remaining product,this is done with the help of "break".

```python
#here we want to find 30 ,once we found the particular number then the loop should stop iterating
nums = [10, 20, 30, 40, 50]

for i in nums:
    if i == 30:
        print("Number found")
        break

    print(i)
```

### continue

**Definition**

"continue" is also an loop control statement.

The "continue" statement is used to skip the current iteration of a loop and move to the next iteration when a condition is met.

Unlike "break", "continue" does not stop the entire loop but it will skip the current iteration and continue with next iteration.

**Example**

Suppose a company has a list of employees and wants to process only active employees.

If an employee is inactive, the program can use "continue" to skip that employee and move to the next employee.

```python
#here we dont want the 3 to be printed so will use continue when loop reaches 3
for number in range(1, 6):
    if number == 3:
        continue    #now here the "3" is skipped and only rest of the numbers are printed

    print(number)
```

### 5. Variable-Length Arguments (*args, **kwargs)
Definition

Variable-length arguments allow a function to accept any number of arguments as input. 

*args:

*args allows a function to accepts any no of non-keyword arguments as input like a tuple.

**kwargs:

while **kwargs allow a function accept any no of keyword arguments as input like a dictionary.

Example:

Suppose we want to create a function to add numbers, but we don't know how many numbers the user will provide. 

*args allows us to pass any number of numbers.

**Real-Time Example**

```python
def add_numbers(*args):   #here we dont know how many nums user wil provide to add,so we are using *args
    total = 0
    for num in args:
        total += num
    return total

print(add_numbers(1, 2, 3))       # 6
print(add_numbers(10, 20, 30, 40)) # 100
```

```python
def show_info(**kwargs):   # same here,we dont know how many student data the user will be storing so we are using kwargs
    for key, value in kwargs.items():
        print(key, ":", value)

show_info(name="Hari", age=21, course="Ece")
```

### 7. Recursive Functions

**Definition**

A recursive function is a function that calls itself to solve a problem. 

It continues calling itself until a specific condition, called the base condition, is reached.

A recursive function solves a big problem by breaking it into smaller sub‑problems of the same type.

Simple Example

Suppose we want to print numbers from 5 down to 1. A function can print one number and then call itself with the next smaller number until it reaches 0.

Real-Time Example

```python
def countdown(n):
    if n == 0:
        return

    print(n)
    countdown(n - 1)

countdown(5)
```

### 2. Generators

**Definition**

A generator is a special type of function that produces result values one at a time instead of creating and storing all the values in memory at once.

Generators use the yield keyword instead of return.

Generators are useful when we have a large amount of data because they save memory by producing values only when they are needed.

Example:

Suppose we want numbers from 1 to 5. Instead of creating the complete list first, a generator can produce 1, then 2, then 3, and so on whenever we ask for the next value.

**Real-Time Example**

Imagine a water dispenser. It does not give all the water at once. It provides water whenever we need it. A generator works in a similar way by producing values only when they are requested.


```python
def g():
    for i in range(5):
        yield i
t=g()
print(next(t))
print(next(t))
print(next(t))
print(next(t))
print(next(t))
print(next(t))  #here when iam training to get the values out of range the StopIteration error is raised

```

### 3. `yield`

**Definition**

The "yield" keyword is used inside a function to produce a result value one at a time and temporarily pause the function. 

When the next value is requested, the function continues from where it stopped.

inside a function the "yield" is used instead of "return"

Example:

Suppose a function needs to produce numbers from 1 to 3. Using `yield`, it can give one number, pause, and continue from the same point when the next number is requested.

**Real Time Example**

A ticket counter can serve one customer at a time. After serving one customer, it waits for the next request instead of serving everyone at once. "yield" works in a similar way by producing one value and pausing.



```python
def get_numbers():
    yield 10
    yield 20
    yield 30

numbers = get_numbers()

print(next(numbers))
print(next(numbers))
print(next(numbers))
```

### 4. Decorators

**Definition**

A decorator is a function that allows us to add extra functionality to another function without changing the original function's code.

Example:

Suppose we already have a function that prints a message. If we want to print something before and after that function runs, we can use a decorator instead of changing the original function.

**Real-Time Example**

A security system can check a person's identity before allowing them to enter a building. The original activity remains the same, but an additional check is added before it. A decorator works in a similar way by adding extra behavior to a function.


```python
def add_message(func):
    def wrapper():
        print("Function is starting")
        func()
        print("Function is completed")
    
    return wrapper


@add_message
def greet():
    print("Hello Hari")


greet()
```

### 5. Closures

**Definition**

A closure is a function it is created when an inner function remembers and uses a value from the outer function even after the outer function has finished executing.

or

A closure is a function that remembers and can access variables from its enclosing (outer) function even after the outer function has finished executing.

Example:

Suppose an outer function creates a message and returns another function. The inner function can still remember the message even after the outer function has finished.

**Real-Time Example**

Imagine creating a personalized notification system. When we create a notification for a particular person, the notification function remembers that person's name and can use it later.



```python
def outer(name): #----this is the outer func
    message = f"Hello, {name}!"
    
    def inner(): #----this is inner func
        print(message)   #inner fnc still remembers 'message' executed from the outer fnc   
    
    return inner

my_func = outer("hariii")
my_func()   #Hello, Hari!
```

### 6. global and nonlocal

**Definition**

The "global" keyword is used when we want to modify a variable that belongs to the global scope. 

The "nonlocal" keyword is used when we want to modify a variable from an enclosing (outer) function.

Example:

A global variable can be accessed by different functions. If we want to change its value inside a function, we can use "global".

Similarly, "nonlocal" allows an inner function to change a variable belonging to its outer function.

**Real-Time Example**

A company may have a global employee count that can be updated whenever a new employee joins. Inside a smaller function, we may also have a local value that needs to be updated by an inner function.


```python
employee_count = 10

def add_employee():
    global employee_count
    employee_count += 1

add_employee()

print("Employees:", employee_count)
```

```python
def counter():
    count = 0

    def increase():
        nonlocal count
        count += 1
        return count

    return increase


counter_function = counter()

print(counter_function())
print(counter_function())
print(counter_function())
```

### 10. Type Hinting

**Definition**


In Python, type hinting means telling the interpreter and other developers what type of data a variable, function parameter, or return value is expected to hold. 

It doesn’t enforce types at runtime cause Python is still dynamically typed, but it improves readability make the code easier to understand. 


Example:

### 1. `map()`

**Definition**

"map()" is used when we want to apply the same operation to every item in a collection. It takes each value one by one, applies a function to it, and gives us the results.

The "map()" function in Python is used to apply a function to each item in an iterable like a list, tuple, or set and return a new iterator with the results.

**Real-Time Example**

Suppose a company has a list of employee ages and wants to find employees who are above 18. `filter()` can be used to keep only those employees who satisfy the condition.



```python
numbers = [10, 15, 20, 25, 30]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
```

### 3. zip()

**Definition**

zip() is used to combine values from two or more collections based on their position. It takes the first value from each collection, then the second value, and so on.

Example:

If we have one list containing names and another list containing ages, zip() can combine the name and age that belong together.

**Real-Time Example**

Suppose a teacher has one list of student names and another list of their marks. zip() can be used to combine each student's name with their corresponding mark.


```python
names = ["Hari", "Rahul", "Kiran"]
marks = [85, 90, 78]

students = list(zip(names, marks))

print(students)
```

### 4. enumerate()

**Definition**

enumerate() function is used when we want to get both the position and the value of each item while looping through a collection.

Example:

If we have a list of fruits, enumerate() can give us the index number along with each fruit.

**Real-Time Example**

Suppose a teacher has a list of students and wants to display their serial numbers along with their names. 
enumerate() can provide both the number and the name.


```python
students = ["Hari", "Rahul", "Kiran"]
 
for index, name in enumerate(students, start=1):  #enumerate() gives two values in every iteration the index and the actual value.

                                                 #I used start=1 so the numbering starts from 1 instead of 0.
    print(index, name)
```
