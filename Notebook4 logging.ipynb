## Logging

### 1. What is Logging?

**Definition**

Logging means keeping a record of what is happening inside a Python program while it is running.

Instead of only looking at the final output, logging allows us to record useful information such as when a program starts, what operation it is performing, whether something went wrong, and when an error happened.

Python provides a built-in logging module for this purpose.

It helps developers track what the program is doing, debug issues, and keep a history of important actions.

simply:

Logging is smarter alternative to print statements.

It gives you structured, leveled, and storable messages about your program’s execution.

Example:

Suppose we have a program that reads a file. We can create log messages when the program starts reading the file, when the file is successfully read, and when an error occurs.

**Real-Time Example**

In a web application, logging can keep track of user requests, successful operations, warnings, and errors. If something goes wrong later, the developer can look at the logs and understand what happened.

```python
import logging  #here this import logging imports Python's built-in logging module.

logging.basicConfig(level=logging.INFO)   #basicConfig() sets the basic logging configuration.

logging.info("Program started")     #logging.info() creates an informational log message.
logging.info("Reading data")
logging.info("Program completed")
```

### 2. Why Use Logging?

**Definition**

Logging is useful because it helps us understand and monitor a program while it is running.

It is especially helpful when the program is large or running on another machine where we cannot constantly watch the output.

Logs can also be saved to a file, which means we can check them later when debugging a problem.

Key reasons to Use Logging:

- Organized messages  
  Logging automatically tags each message with its level (INFO, WARNING, ERROR, etc.), so you instantly know how serious it is.

- Saves to files  
  Unlike print(), logs can be written to a file for later review — perfect for tracking issues in long‑running or production programs.

- Filter by importance  
  You can configure logging to show only certain levels (e.g., only errors), which keeps output clean and focused.

- Professional debugging  
  Logging is the industry standard — it’s how real applications track what’s happening behind the scenes.

- Flexible output  
  Logs can go to console, files, or even external systems (like monitoring tools), making them far more versatile than print().

Example:

Instead of printing every step of a program, we can use logging to record important events such as starting a process, completing a process, or finding an error.

**Real-Time Example**

Suppose an ML model is running on a server and suddenly stops working. We may not be sitting in front of the server when the problem happens. If the application has logging, we can check the log file and find where the problem occurred.

```python
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started")  #here the program creates a log when it starts.

try:
    result = 10 / 2
    logging.info("Calculation completed successfully") #After the calculation is completed, another log message is created.

except Exception as error:
    logging.error("An error occurred: %s", error) #If an error occurs, the except block records the error using logging.error().
```

### 3. print() vs Logging

**Definition**

print() is mainly used to display something directly on the screen or console.

Logging is designed for recording information about what is happening inside an application.

Logging gives us more control because we can use different levels, save messages to files, include timestamps, and decide where the messages should be sent.

For larger applications, logging is generally more useful because the messages can be filtered, formatted, and stored.

Example:

With print(), we simply display a message.

With logging, we can specify whether the message is normal information, a warning, or an error.

**Real-Time Example**

While practicing Python, print() may be enough to check a variable. But in a real application, logging is more useful because developers may need to keep track of errors and application activity even after the program has finished.

```python
import logging

logging.basicConfig(level=logging.INFO)

print("Program started")  #print() simply displays the message.

logging.info("Program started")
logging.warning("This is a warning")
logging.error("Something went wrong")   #Logging provides additional information about the importance of the message.
```

### 4. Log Levels

**Definition**

Log levels are used to indicate how important a particular log message is.

Log levels let you filter messages depending on importance.

Python provides different levels so that we can separate normal information from warnings and serious errors.

The commonly used levels are:

- DEBUG (10) → Detailed information for developers, like step‑by‑step flow or variable values.

- INFO (20) → General events confirming the program is working as expected.

- WARNING (30) → Something unexpected happened, but the program can still run.

- ERROR (40) → A serious problem occurred; part of the program failed.

- CRITICAL (50) → Very serious error; program may crash or stop running.

Each level represents a different level of importance.

Example:

A message saying "Program started" can be INFO, while a message saying "Database connection failed" can be ERROR.

**Real-Time Example**

In a production application, we may have thousands of log messages. Log levels help developers quickly focus on warnings and errors instead of reading every normal message.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Debug message")
logging.info("Information message")
logging.warning("Warning message")
logging.error("Error message")
logging.critical("Critical message")
```

Each logging function represents a different level.

Setting the level to DEBUG allows all these messages to be displayed.

The higher levels represent more serious situations.

### 5. DEBUG

**Definition**

DEBUG is used for detailed information that is mainly useful while developing or troubleshooting a program.

It’s used for detailed diagnostic information things only developers usually care about.

Think of it as a microscope: it shows step‑by‑step flow, variable values, and internal details.

These messages can help us understand what is happening internally.

They are usually not necessary for normal users.

Use DEBUG when you want to trace exactly what’s happening inside your program.

Example:

We can use DEBUG to record the value of a variable or to know which step of a function is currently running.

**Real-Time Example**

While debugging an ML program, we may want to know the number of rows loaded, the shape of the dataset, or which processing step is currently running. These details can be logged using DEBUG.

```python
import logging

logging.basicConfig(level=logging.DEBUG)

name = "Hari"
age = 22

logging.debug("Name value: %s", name)
logging.debug("Age value: %s", age)
```

### 6. INFO

**Definition**

INFO is used for normal messages that tell us what the application is doing.

It is useful for recording successful operations and important steps in the program.

It’s used to record normal events that confirm the program is running as expected.

Think of it as a status update — not too detailed like DEBUG, but important enough to know the program’s progress.

Example:

We can use INFO when a program starts, when a file is loaded successfully, or when a database connection is established.

**Real-Time Example**

In a data pipeline, we can log messages such as "Data loading started", "1000 records loaded", and "Data processing completed" using the INFO level.

```python
#Use INFO for general updates about the program’s normal operation.
import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started")
logging.info("Data loaded successfully")
logging.info("Processing completed")
```

### 7. WARNING

**Definition**

WARNING is used when something unexpected has happened, but the program can still continue running.

or

It signals that something unexpected happened, but the program can still continue running.

It tells us that we should pay attention to a situation.

Example:

If a program is using a default value because some data was missing, we can create a warning.

**Real-Time Example**

Suppose an ML dataset has some missing values. The program can continue processing the data, but a warning can tell the developer that some values were missing.

```python
#here  the age value is missing but the program can still  run 
import logging

logging.basicConfig(level=logging.WARNING)

age = None

if age is None: #The program checks whether age has a value.
    logging.warning("Age value is missing")  #Since it is None, a warning is generated.
```

### 8. ERROR

**Definition**

ERROR is used when something has gone wrong and a particular operation could not be completed.

The application may still continue running depending on how the error is handled.

It indicates a serious problem that has occurred — something failed in the program.

Example:

If a file cannot be opened or a database query fails, we can record the problem using ERROR.

**Real-Time Example**

If an application cannot connect to its database, we can record an error message so that the developer knows that the database operation failed.

```python
import logging

logging.basicConfig(level=logging.ERROR)

try:
    number = 10 / 0

except ZeroDivisionError as error:
    logging.error("Calculation failed: %s", error)
```

### 9. CRITICAL

**Definition**

CRITICAL is used for very serious problems that may prevent the application from continuing normally.

It represents a more serious situation than a normal ERROR.

It signals a very serious error — something that may cause the program to crash or stop running entirely.

Example:

If an important service required by the application completely fails, we can record it as a critical problem.

**Real-Time Example**

Suppose a production application completely loses its database connection and cannot perform any important operation. This could be recorded as a CRITICAL message because the application may not be able to continue normally.

```python
import logging

logging.basicConfig(level=logging.CRITICAL)

logging.critical("Database server is completely unavailable")
```

### 10. Log Formatters

**Definition**

A log formatter controls how a log message is displayed.

By default, logging may only show the level and message. With a formatter, we can include useful information such as the date, time, log level, and message.

Example:

Instead of seeing only:

INFO: Data loaded

we can make the log look something like:

2026-09-25 10:30:15 - INFO - Data loaded

**Real-Time Example**

When checking logs from a production application, the time of the event is very useful. If an error happened at 10:30 AM, the developer can use the timestamp to match it with other events.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")
logging.warning("Low memory available")
logging.error("File could not be opened")
```

format decides how every log message should look.

%(asctime)s → time of the log
%(levelname)s → log level
%(message)s → actual message

This makes logs much easier to understand.

### 11. File Handlers

**Definition**

A FileHandler is a logging component that writes log messages into a file instead of just showing them on the console.

A file handler is used when we want to save log messages into a file instead of only displaying them on the screen.

This is useful because we can open the log file later and check what happened in the application.

Example:

We can create a file called app.log and store our application's log messages inside it.

**Real-Time Example**

A web application running on a server can save its errors and important events into a log file. If something goes wrong during the night, the developer can check the file the next morning.

```python
import logging

# Create a logger
logger = logging.getLogger("my_logger")
logger.setLevel(logging.DEBUG)

# Create a FileHandler
file_handler = logging.FileHandler("app.log")

# Set a format for log messages
formatter = logging.Formatter("%(levelname)s - %(message)s")
file_handler.setFormatter(formatter)

# Add the handler to the logger
logger.addHandler(file_handler)

# Log some messages
logger.debug("Debugging details")
logger.info("Program started")
logger.error("Something went wrong")
```

### 12. Console Handlers

**Definition**

A console handler is used to send log messages to the console or terminal where the program is running.

or

A ConsoleHandler is a logging handler that outputs logs directly to the console (stdout).

It is useful when we want to see what is happening in the application immediately.

Why ConsoleHandler Is Useful:

- Customization → You can control the format, level, and destination of console logs separately from file logs.

- Multiple outputs → You might want logs to go to both a file and the console. Handlers let you do that.

- Fine‑grained control → You can set different severity levels for console vs file.

   For example:

   Console shows only WARNING and above (to avoid clutter).

   File saves everything including DEBUG.

- Flexibility → In larger apps, you often need multiple handlers (console, file, network, etc.), and ConsoleHandler is the clean way to manage console   output.

Example:

Instead of saving a message into a file, we can configure a StreamHandler to display it in the terminal.

**Real-Time Example**

While developing an application in VS Code, we may want to immediately see whether a database connection was successful or whether an error occurred. A console handler can display those messages directly in the terminal.

```python
import logging

# Create a logger
logger = logging.getLogger("my_logger")
logger.setLevel(logging.DEBUG)

# Create a ConsoleHandler
console_handler = logging.StreamHandler()

# Set a formatter for console output
formatter = logging.Formatter("%(levelname)s - %(message)s")
console_handler.setFormatter(formatter)

# Add handler to logger
logger.addHandler(console_handler)

# Log messages
logger.debug("Debugging details")
logger.info("Program started")
logger.warning("Low disk space")
logger.error("File not found")
logger.critical("System crash imminent")
```

### 13. Rotating Logs (Bonus)

**Definition**

Rotating logs means automatically creating a new log file when the current log file becomes too large.

This prevents one log file from growing forever and taking up too much disk space.

Old log files can be kept as backups.

Example:

Suppose an application generates a lot of logs every day. Instead of keeping everything in one huge file, we can rotate the file after it reaches a certain size.

- RotatingFileHandler → rotates logs based on file size.

    Example: when app.log reaches 5 MB, it’s renamed to app.log.1, and a new app.log starts.

    You can set how many backups to keep.

- TimedRotatingFileHandler → rotates logs based on time.

    Example: create a new log file every midnight, or every hour.

    Old logs are archived with timestamps.

**Real-Time Example**

A busy web application may generate thousands of log messages every day. If all of them are stored in one file, the file can become very large. Rotating logs helps divide the logs into smaller files.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("rotating_app")
logger.setLevel(logging.INFO)

handler = RotatingFileHandler(
    "app.log",
    maxBytes=1000,
    backupCount=3
)

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

handler.setFormatter(formatter)

logger.addHandler(handler)

for i in range(100):
    logger.info("Processing record %s", i)
```

### 14. Logging Best Practices

**Definition**

Logging best practices are simple rules that help us create useful and manageable logs.

Good logging should give enough information to understand what happened without filling the log with unnecessary messages.

Best Practices
- Use logging, not print

   print() is fine for quick debugging, but logging is structured, filterable, and professional.

- Choose the right log level

   Use DEBUG for detailed developer info, INFO for normal events, WARNING for unusual situations, ERROR for failures, and CRITICAL for severe crashes.

- Configure handlers properly

   Use FileHandler to save logs, ConsoleHandler for live output, and RotatingFileHandler to prevent huge log files.

- Set a consistent format

   Include timestamp, level, and message. Example:
   logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")

  
- Avoid logging sensitive data

   Don’t log passwords, personal info, or API keys.

Example:

Instead of writing:

logging.info("Something happened")

we can write something more useful:

logging.info("Successfully loaded 500 records")

The second message gives us more information about what actually happened.

**Real-Time Example**

In a real application, developers may have to check logs when something goes wrong. Clear messages, proper log levels, timestamps, and useful error details make it much easier to find the problem.

```python
import logging

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s"
)

logging.info("Application started")

try:
    result = 100 / 5
    logging.info("Calculation completed successfully")

except ZeroDivisionError:
    logging.error("Division by zero occurred")

logging.info("Application finished")
```

This example follows some basic logging practices:

- Use the correct log level for each situation.
- Include useful information in the message.
- Include timestamps using a formatter.
- Use ERROR when something actually goes wrong.
- Avoid logging unnecessary information.
- Do not use logging messages as a replacement for proper error handling.
