## File Handling 

### 1. Reading Text Files

**Definition**

Reading a text file means opening a file stored on your computer and loading its contents into your Python program so you can process or display the text.

or

Reading a text file is just opening it in "r" (read mode) and using methods like read(), readline(), or readlines() to get the content.

Python provides the open() function for working with files.

The most common mode for reading a text file is "r".

- file.read() → reads the whole file as one string.

- file.readline() → reads one line at a time.

- file.readlines() → reads all lines into a list.

Example:

Suppose we have a file called notes.txt containing:

Python is easy to learn.
File handling is useful.

Python can open this file and read its contents.

**Real-Time Example**

Think about a Python application that stores customer information in a text file. The application can read that file whenever it needs to display the stored information.

```python
file = open("student.txt", "r")  #here open("notes.txt", "r") opens the file in read mode

data = file.read()  #file.read() reads the complete content of the file.

print(data) 

file.close()  #here we should manully close the file
```

```python
# Automatically closes the file after use
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

### 2. Writing Text Files

**Definition**

Writing a file means adding data to a file using Python.

We can use the open() function with "w" mode to write data into a text file.

If the file does not exist, Python creates it automatically.

If the file already exists, "w" mode removes the old content and writes the new content.

- "w" → write (overwrites file if it exists).

- "a" → append (adds new content at the end without deleting old content).

- "x" → exclusive creation (fails if file already exists).

Example:

If we want to create a file called student.txt and store a student's information, we can use write mode.

**Real-Time Example**

A Python application can create a report file and write the generated results into that file.

```python
file = open("student.txt", "w")  #"w" means write mode.

file.write("Name: Hari\n") #file.write() writes data into the file.
file.write("Age: 22\n")     #\n moves the next text to a new line.
file.write("Course: Python")

file.close()
```

### 3. Appending Data to a File

**Definition**

Appending means adding new data to the end of an existing file without removing the old data.

Python uses "a" mode for appending.

Example:

Suppose student.txt already contains:

Name: Hari
Age: 22

If we use append mode, we can add another line without deleting the existing information.

**Real-Time Example**

A logging system can append new log messages to the same file instead of deleting the previous logs.

```python
file = open("student.txt", "a")

file.write("\nCourse: Python")

file.close()
```

### 4. CSV Files

**Definition**

A CSV file is commonly used to store tabular data in rows and columns.

A CSV file stands for Comma-Separated Values file. It’s one of the simplest formats used to store tabular data (like spreadsheets or databases) in plain text.

or

A CSV file is a plain text file that stores tabular data using commas to separate values.

A CSV file contains data separated by commas (,).

Each line in the file represents a row.

Each value in the line represents a column.

It’s widely used because it’s lightweight, human-readable, and supported by almost all data tools (Excel, databases, Python, etc.).

example:

id,name,age
1,Hari,22
2,Ram,23
3,Ravi,21

Each line represents a row, and commas separate the values.

Python provides the built-in csv module to work with CSV files.

Example:

A student database can be stored in a CSV file.

**Real-Time Example**

Data analysts often receive datasets in CSV format and use Python with Pandas to read and process the data.

```python
#reading a csv file
import csv  #here import csv imports Python's CSV module

with open("students.csv", "r") as file:  #open() opens the CSV file
    reader = csv.reader(file)   #csv.reader() reads the rows.

    for row in reader:
        print(row)
```

```python
#writing to a csv file
import csv

with open("students.csv", "w", newline="") as file: #here we r opening file in write mode
    writer = csv.writer(file)  #csv.writer() is used to write rows into a CSV file.

    writer.writerow(["id", "name", "age"])  #writerow() writes one complete row at a time.
    writer.writerow([1, "Hari", 22])
    writer.writerow([2, "Ram", 23])

```

### 5. JSON Files

**Definition**

JSON stands for JavaScript Object Notation.

JSON is a text-based format commonly used for storing and exchanging structured data.

A JSON file is a plain text file that stores data in the JavaScript Object Notation (JSON) format. 

It’s widely used for data exchange between applications because it’s lightweight, human‑readable, and easy for machines to parse.

JSON data is written using structures similar to Python dictionaries and lists.

Example:

{
    "name": "Hari",
    "age": 22,
    "skills": ["Python", "SQL", "Machine Learning"]
}

**Real-Time Example**

When a Python application communicates with an API, the data received from the API is very often in JSON format.

```python
#here we r reading a json file 
import json ##import json imports the JSON module.

with open("student.json", "r") as file:
    data = json.load(file)   #json.load() reads JSON data from the file and converts it into a Python object, usually a dictionary or list.

print(data)
```

```python
#here we r writing to a json file
import json  #import json imports the JSON module.

student = {
    "name": "Hari",     #student is a Python dictionary.
    "age": 22,
    "course": "Python"
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)  #json.dump() writes the Python dictionary into the JSON file.
```

### 6. Pickle Files

**Definition**

Pickle is a Python module used to serialize and deserialize Python objects.

A Pickle file in Python is a special type of file that stores Python objects in a serialized (binary) format. 

The process is called pickling (saving) and unpickling (loading).

Serialization means converting a Python object into a format that can be stored in a file.

Deserialization means loading that stored object back into Python.

Unlike JSON, Pickle can store many Python-specific objects.

**Real-Time Example**

A machine learning model can be saved using Pickle so that the trained model can be loaded later without training it again.

```python
#serialization
import pickle

student = {    #this is our python object
    "name": "Hari",
    "age": 22,
    "course": "AI/ML"
}

with open("student.pkl", "wb") as file:   #"wb" means write binary.
    pickle.dump(student, file)  #pickle.dump() saves the Python object into the file.
```

```python
#deserialization
import pickle

with open("student.pkl", "rb") as file:   #Here "rb" means read binary.
    data = pickle.load(file)

print(data)
```

### 7. Context Manager (with)

**Definition**
A Context Manager is a Python construct that defines setup and cleanup actions around a block of code.

A context manager is a way of managing resources automatically.

It ensures resources are properly managed (opened, used, then closed) without you having to manually handle them.

When working with files, the with statement automatically closes the file after the block of code finishes.

Because of this, we normally prefer with open() instead of manually opening and closing files.

```python
# example Without "with":

file = open("notes.txt", "r")

data = file.read()

file.close()
```

```python
# example With "with":

with open("notes.txt", "r") as file:
    data = file.read()

print(data)
```

### 8. pathlib

**Definition**

pathlib is a Python module used to work with file and directory paths.

It provides the Path class, which makes path handling easier and more readable than manually creating paths using strings.

Instead of treating paths as plain strings, it represents them as Path objects with useful methods and properties.

few methods:

- .exists() → check if path exists

- .is_file() / .is_dir() → check type

- .read_text() / .write_text() → quick file operations

- .iterdir() → list directory contents

**Real-Time Example**

A machine learning project may have separate folders for datasets, models, logs, and output files. pathlib can be used to create and manage these paths.

```python
from pathlib import Path

# Create a Path object
path = Path("student.txt")

# Check if file exists
print(path.exists())   # True/False

# Get absolute path
print(path.resolve())

# Read text from file
print(path.read_text())

# Write text to file
path.write_text("Hello from pathlib!")
```

### 9. os

**Definition**

os is a built-in Python module that allows us to interact with the operating system.

It can be used for working with files, directories, environment variables, paths, and other operating-system-related operations.

It lets Python programs perform system-level tasks like creating folders, deleting files, checking paths, or reading environment variables.

Key Features:

- File & Directory Management → create, delete, rename, list.

- Path Handling → combine paths (os.path.join), check existence (os.path.exists).

- Environment Variables → access system settings like PATH.

- Process Management → run system commands (os.system("dir")).

**Real-Time Example**

A Python application can use os to check the current directory, create folders, or get information about files.

```python
import os

# Current working directory
print(os.getcwd())

# Change directory
os.chdir("C:/Users/Hari/Documents")

# List files in a directory
print(os.listdir("."))

# Create a new folder
os.mkdir("new_folder")

# Remove a file
os.remove("example.txt")

# Environment variables
print(os.environ["PATH"])
```

### 10. shutil

**Definition**

shutil stands for shell utilities.

It is a Python module used for higher-level file and directory operations such as copying, moving, and deleting directories..

Provides functions for copying, moving, deleting files and directories.

Useful when you need to manipulate files beyond just opening/reading/writing.

Key Features:

- File copy → copy(), copy2()

- Directory copy → copytree()

- Move files/folders → move()

- Delete directories → rmtree()

- Disk usage info → shutil.disk_usage(path)

- Archive creation/extraction → shutil.make_archive() and shutil.unpack_archive()

**Real-Time Example**

A backup application can use shutil to copy important files from one folder to another folder.

```python
import shutil

# Copy a file
shutil.copy("source.txt", "destination.txt")

# Copy a file with metadata (permissions, timestamps)
shutil.copy2("source.txt", "destination.txt")

# Copy an entire directory
shutil.copytree("source_folder", "destination_folder")

# Move a file or folder
shutil.move("old_location/file.txt", "new_location/file.txt")

# Delete an entire directory
shutil.rmtree("folder_to_delete")
```

### 11. File Validation

**Definition**

File validation means checking whether a file is available and whether it is suitable before using it.

or

File Validation in Python means checking whether a file meets certain conditions before you use it — this prevents errors and ensures your program works safely.

For example, we may check:

Does the file exist?
Is it actually a file?
Does it have the correct extension?
Is the file empty?
Is the file size reasonable?

**Real-Time Example**

Before loading a CSV dataset for a machine learning project, we can check whether the file exists and whether it has the expected .csv extension.

```python
from pathlib import Path

file_path = Path("students.csv")

if not file_path.exists():
    print("File does not exist")

elif not file_path.is_file():
    print("This is not a file")

elif file_path.suffix != ".csv":
    print("Invalid file type")

elif file_path.stat().st_size == 0:
    print("File is empty")

else:
    print("File is valid")
```

### 12. Directory Operations

**Definition**

A directory is another name for a folder.

Directory operations are actions such as creating, checking, listing, renaming, and deleting folders.

Python provides several ways to work with directories using modules such as os, pathlib, and shutil.

**Real-Time Example**

A data science project may automatically create folders such as data, models, logs, and output when the project starts.

```python
#folder creation with pathlib
from pathlib import Path

folder = Path("new_folder")

folder.mkdir(exist_ok=True)

print("Directory created")
```

```python
#folder creation with os module
import os   # Always import the module first

# Create a single folder
os.mkdir("new_folder")

# Create nested folders
os.makedirs("parent/child/grandchild")
```

```python
#checking whether the folder is empty or not using path lib
from pathlib import Path

folder = Path("new_folder")
if not any(folder.iterdir()):
    print("empty")
else:
    for file in folder.iterdir():
        print(file)
```

```python
##checking whether the folder is empty or not using os module
import os

folder = "new_folder"

# Check if folder exists
if os.path.exists(folder):
    # Check if folder is empty
    if not os.listdir(folder):   # returns [] if empty
        print("empty")
    else:
        for file in os.listdir(folder):
            print(file)
else:
    print("Folder does not exist")
```
